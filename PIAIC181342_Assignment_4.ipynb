## ***Exercise 3-7: Shrinking Guest List***
You just found out that your new dinner table won’t arrive in time for the dinner, and now you have space for only two guests.

* Start with your program from Exercise 3-6. Add a new line that prints a message saying that you can invite only two people for dinner.
* Use pop() to remove guests from your list one at a time until only two names remain in your list. Each time you pop a name from your list, print a message to that person letting them know you’re sorry you can’t invite them to dinner.
* Print a message to each of the two people still on your list, letting them know they’re still invited.
* Use del to remove the last two names from your list, so you have an empty list. Print your list to make sure you actually have an empty list at the end of your program.


```python
print("\nUnfortunately, the new dinner table won’t arrive in time. We can invite only two people.")

# Removing guests
while len(guests) > 2:
    removed_guest = guests.pop()
    print(f"Sorry, {removed_guest}, we can’t invite you to dinner.")

# Notifying the remaining guests
print("\nFinal invitations:")
for guest in guests:
    print(f"Dear {guest}, you are still invited to dinner.")

# Clearing the list
del guests[:]
print("\nFinal guest list (should be empty):", guests)
```

Think of at least five places in the world you’d like to visit.

* Print your list in its original order. Don’t worry about printing the list neatly; just print it as a raw Python list.
*  Use sorted() to print your list in alphabetical order without modifying the actual list.
*  Show that your list is still in its original order by printing it.
* Use sorted() to print your list in reverse-alphabetical order without changing the order of the original list.
* Show that your list is still in its original order by printing it again.
* Use reverse() to change the order of your list. Print the list to show that its order has changed.
* Use reverse() to change the order of your list again. Print the list to show it’s back to its original order.
* Use sort() to change your list so it’s stored in alphabetical order. Print the list to show that its order has been changed.
* Use sort() to change your list so it’s stored in reverse-alphabetical order.
 Print the list to show that its order has changed.

```python
places = ['Lahore', 'Islamabad', 'Londan', 'Sydney', 'New York']

print("\nOriginal list of places:", places)
print("Alphabetical order:", sorted(places))
print("Original list (unchanged):", places)
print("Reverse alphabetical order:", sorted(places, reverse=True))
print("Original list (unchanged):", places)

places.reverse()
print("List reversed:", places)
places.reverse()
print("List restored:", places)

places.sort()
print("List sorted alphabetically:", places)
places.sort(reverse=True)
print("List sorted in reverse alphabetical order:", places)
```

## ***Exercise 3-10: Every Function***
* Think of things you could store in a list. For example, you could make a list of mountains, rivers, countries, cities, languages, or anything else you’d like. Write a program that creates a list containing these items and then uses each function introduced in this chapter at least once.



```python
items = ['Mount Everest', 'Amazon River', 'Japan', 'Paris', 'Python']

print("\nOriginal list of items:", items)
items.append('Sahara Desert')
print("After appending an item:", items)

items.insert(2, 'Grand Canyon')
print("After inserting an item:", items)

items.remove('Japan')
print("After removing an item:", items)

print("Using sorted():", sorted(items))
items.sort()
print("Sorted permanently:", items)
items.reverse()
print("Reversed list:", items)
print("Length of the list:", len(items))
```
