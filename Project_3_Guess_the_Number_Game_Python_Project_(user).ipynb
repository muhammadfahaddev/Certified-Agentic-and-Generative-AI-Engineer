```python
import random
import time
```

```python

def GuessUser():
  print("Guess a number between 1 and 500")
  randomNumber=random.randint(1,500)
  attempt=8
  while attempt>0:
    try:
      start_time=time.time();
      userGuess=int(input("Enter your guess number within 8 seconds: "))
      if time.time()-start_time>8:
         print("too slow, you lost an attempt");
      else:
        if(randomNumber==userGuess):
          print(f"Your guess is correct {userGuess}")
          return;
        elif abs(randomNumber-userGuess)<=10:
          print("Your guess is close to the number")
        elif randomNumber<userGuess:
          print("Too High")
        elif randomNumber>userGuess:
          print("Too Low")
        attempt=attempt-1;
        print(f'Your attempt is available now {attempt}');

        if attempt==0:
          print('You are lost ');
          return;
    except ValueError:
      print("Invalid input. Please enter a valid number.")
```
