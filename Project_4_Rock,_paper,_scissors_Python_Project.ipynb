```python
import random
```

```python
def GuessGame():
    guessList = ['rock', 'paper', 'scissor']
    userCount=0
    computerCount=0
    print("\nWelcome to Rock-Paper-Scissors!")

    while True:
        guess = random.choice(guessList)
        user_guess = input("\nChoose Rock, Paper, or Scissor or type 'exit' to quit: ").lower()
        if user_guess == 'exit':
            print("Thanks for playing! Goodbye.")
            break
        if user_guess not in guessList:
            print("Invalid choice! Please enter Rock, Paper, or Scissor ")
            continue


        print(f"Computer chose: {guess}")
        if user_guess == guess:
            print("It's a tie! Try again.")
            continue
        if (user_guess == "rock" and guess == "scissor") or \
           (user_guess == "scissor" and guess == "paper") or \
           (user_guess == "paper" and guess == "rock"):
            print("You Won!")
            userCount+=1
        else:
            print("You Lost! Try again.")
            computerCount+=1

        print(f"Your Score: {userCount}")
        print(f"Computer's Score: {computerCount}")

    if userCount<computerCount:
      print("Computer Won!")
    elif userCount>computerCount:
      print("You Won!")
    elif userCount==computerCount==0:
      print("Game Quit!")
    else:
      print("It's a tie!")
```

```python
GuessGame()
```
