```python
import google.generativeai as genai
from google.colab import userdata
```

```python
API_KEY=userdata.get('API_KEY')
genai.configure(api_key=API_KEY)
```

```python
def GenerateMadLibsGame():
  print("Welcome to Mad Libs Game\n");
  noun=input("Enter Noun Word: ");
  adjective=input("Enter adjective Word: ");
  verb=input("Enter Verb Word: ");
  place=input("Enter place Word: ");
  adverb=input("Enter adverb Word: ");

  promatText=f"""
  Create funny Mad Libs Story using the following words:
  - Noun:{noun},
  - Adjective:{adjective},
  - Verb:{verb},
  - place:{place},
  - Adverb:{adverb},
  Make it funny and short and easy wording story.
  """;
  modal=genai.GenerativeModel("gemini-1.5-flash");
  response=modal.generate_content(promatText)
  print("\nGenerated Mad Libs Story:\n");
  return response.text;
```

```python
result=GenerateMadLibsGame();
print(result)
```
