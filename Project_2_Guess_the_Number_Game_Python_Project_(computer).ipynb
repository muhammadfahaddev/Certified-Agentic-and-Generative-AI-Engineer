```python
import random
```

```python
low,high=1,100;
attempt=5;
feedback='';

print("\nGuess the Number Game\n");
print(f"Guess a number between {low} and {high}");

while feedback !='c' and attempt>0:
  randomNumber = random.randint(low, high)

  feedback=input(f"Is {randomNumber} if too low than select (L) or if too high than select (H) and if Correct than Select (C): ").lower();

  if feedback=='h':
    high=randomNumber-1;
  elif feedback=='l':
    low=randomNumber+1;

  attempt=attempt-1;
  print(f"You have {attempt} attempts left");

if feedback == 'c':
    print(f"\nYas! I guessed your number {randomNumber} correctly!")
else:
  print("Oh no! I ran out of attempts. You win!")



```
